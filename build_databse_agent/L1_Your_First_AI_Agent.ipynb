```python
# 1. Leverage Langchain
# azure endpoint and azure api key will get loaded automatically

from dotenv import load_dotenv
from langchain_core.messages import HumanMessage
from langchain_openai import AzureChatOpenAI


load_dotenv(override=True)

model = AzureChatOpenAI(
    api_version=os.getenv("AZURE_OPENAI_API_VERSION"),
    azure_endpoint=os.getenv("AZURE_OPENAI_ENDPOINT"),
    azure_deployment=os.getenv("AZURE_OPENAI_DEPLOYMENT")
)
```

```python
# prepare prompt 
message = HumanMessage(content="Translate this sentence from English to French " 
"and Spanish. I like red cars and blue houses, but my dog is yellow.")
```

```python
# engage the model to receive a response
model.invoke([message])
```
